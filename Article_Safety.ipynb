В последние годы большие языковые модели (LLM) стали основой множества приложений — от чат-ботов и поиска до генерации кода и анализа данных. Однако рост их возможностей сопровождается и ростом рисков: злоумышленники активно исследуют способы обхода встроенных ограничений, извлечения конфиденциальной информации и манипулирования поведением моделей. Особое внимание сегодня привлекают LLM-агенты — автономные системы, которые используют языковые модели для планирования и выполнения действий через различные инструменты и API. Такие агенты значительно расширяют поверхность атаки: успешный взлом может привести не только к некорректному тексту, но и к реальным действиям — от утечки данных до несанкционированных транзакций.

Таксономия атак = Механизмы × Каналы обхода (**двумерная классификация**):

| **Механизмы контроля (C)** | **Каналы обхода (E)** | **Пример** |
|-----------------------------|-----------------------|-------------|
| **C1. Модификация цели** (изменение функции потерь) | **E1. Контекст/состояние** (роли, перефразирование, многоязычность) | Добавление штрафа за токены отказа (`Don’t Say No`) |
| | **E2. Кодирование/поверхностные изменения** (символьные замены, шифрование) | Замена символов (`"hack" → "h@ck"`) |
| | **E3. Визуальный канал** (изображения + текст) | Внедрение вредоносного промта в картинку |
| **C2. Методы обновления** (оптимизация поиска) | **E1–E3** | Координатный градиентный спуск (GCG), импульсные методы (MAC) |
| **C3. Поиск с оценщиком** (использование внешних моделей для ранжирования) | **E1–E3** | Генетические алгоритмы (AutoDAN), лучевой поиск (RIPPLE) |
| **C4. Манипуляция представлениями** (вмешательство в скрытые состояния) | **E1–E3** | Прямое изменение логитов токенов (JAILMINE) |
| **C5. Обучение политики** (RL, обучение генераторов атак) | **E1–E2** | Многоагентный RL (PathSeeker), обучение эмбеддингов (ASETP) |

Подробнее по каналам обхода
 1. E1. Манипуляция контекстом/состоянием
- Изменение ролей, перефразирование, вложенные инструкции.

 2. E2. Кодирование/поверхностные преобразования
- Изменение символов, токенов, синтаксиса, языка, сохраняя смысл.

 3. E3. Мультимодальные атаки
- Использование визуального/аудио канала для обхода текстовой защиты.

Подробнее по механизмам контроля
 1. C1. Модификация цели
- Изменяют функцию потерь, чтобы уменьшить вероятность отказа или увеличить вероятность целевого ответа (например, Don’t Say No — штрафует токены, связанные с отказом).

 2. C2. Методы обновления
- Улучшают процесс поиска атак в дискретном пространстве токенов (Gradient-based, Momentum-based обновления токенов).

 3. C3. Поиск с оценщиком (Evaluator-Guided)
- Используют внешнюю модель-оценщик (например, LLM-as-judge), чтобы ранжировать кандидатов.

 4. C4. Манипуляция представлениями
- Прямое вмешательство в скрытые состояния или логиты модели.

 5. C5. Обучение политики
- Обучают генеративные модели или агентов для автоматического создания атак.
