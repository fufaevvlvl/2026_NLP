1.

В последние годы большие языковые модели (LLM) стали основой множества приложений — от чат-ботов до генерации кода. Однако рост их возможностей сопровождается ростом рисков: злоумышленники активно исследуют способы обхода встроенных ограничений, извлечения конфиденциальной информации и манипулирования поведением моделей. Особое внимание сегодня привлекают LLM-агенты — автономные системы, которые используют языковые модели для планирования и выполнения действий: успешный взлом может привести не только к некорректному ответу, но и к реальным действиям, например, несанкционированным транзакциям.

Статья A Survey of Jailbreaking Attacks on Large Language and Multimodal Models https://dl.acm.org/doi/epdf/10.1145/3789982.3790001 выпущена по материалам конференции  8th Artificial Intelligence and Cloud Computing Conference 2025 (AICCC). Она представляет из себя масштабное исследование и предлагает систематическую классификацию методов защиты и взлома LLM, а также содержит подборку датасетов для калибровки механизмов защиты.


3.

Классификация атак представляет собой комбинацию механизма контроля (C1-C4) и канала обхода (E1-E3).

 C1. **C1. Модификация цели**. Изменение функции потерь, например, Don’t Say No в промпте - штрафуем модель за отказы.

 C2. **C2. Методы обновления**. Gradient-based, Momentum-based обновления токенов и промптов.

 C3. **C3. Поиск с оценщиком**. LLM-as-judge, чтобы ранжировать кандидатов.

 C4. **C4. Манипуляция представлениями**. Прямое вмешательство в скрытые состояния или логиты модели.

 C5. **C5. Обучение политики**. Обучение генеративных моделей или агентов для автоматического создания атак.

 E1. **E1. Контекст/состояние**. Изменение ролей, перефразирование, вложенные инструкции.

 E2. **E2. Кодирование/поверхностные изменения**. Изменение символов, токенов, синтаксиса, языка.

 E3. **E3. Мультиканальность**. Использование визуального/аудио/Web канала.
